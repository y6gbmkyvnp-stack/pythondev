```python
class Baseball_Team:
    def __init__(self, name, win, lose, draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw

    def calc_win_rate(self):
        #　勝率　= 勝ち数　/ (勝ち数 + 負け数)
        if self.win + self.lose == 0:
            return 0
        return self.win / (self.win + self.lose)
       
    def show_team_result(self):
        # 表の1行分を表示する   
       print(f"{self.name:<10}"
                f"{self.win:>3}"
                f"{self.lose:>3}"
                f"{self.draw:>3}"
                f"{self.calc_win_rate():.3f}")

```

```python
# 表の見出しを表示
print("team   win lose draw tate")

team = [
    Baseball_Team("Archers", 77, 64, 2),
    Baseball_Team("Meteors", 71, 69, 3),
    Baseball_Team("Cheetahs", 69, 68, 6),
    Baseball_Team("Sharks", 70, 70, 3),
    Baseball_Team("Pysons", 68, 73, 2),
    Baseball_Team("Owls", 59, 82, 2),
]

# 各チームの成績を表示
for team in team:
    team.show_team_result()
```
