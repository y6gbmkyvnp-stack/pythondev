```python
import string
#英単語を格納するリスト
words = []

# 英単語を繰り返し入力
while True:
    word = input("英単語を入力してください")
    if word == "":
        break
    words.append(word)

words.sort()
print(f"入力した英単語:{words}")#入力した英単語を表示

chars =[]
for word in words:
    chars.extend(word)

for ch in string.ascii_lowercase:
    count = chars.count(ch)
    if count > 0:
        print(f"{ch}:{count}個ありました")




```
